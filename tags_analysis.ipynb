```python
import json
from collections import Counter
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
with open('dataset/arctic-0-85-1.0.0.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

all_tags = [tag for task in data['tasks'] for tag in task['tags']]
tag_counts = Counter(all_tags)
print(f"Tasks: {len(data['tasks'])}")
print(f"Unique tags: {len(tag_counts)}")
```

```python
len(data['tasks'])
```

```python
difficulty_counts = {tag: tag_counts[tag] for tag in ('easy', 'medium', 'hard', 'expert')}
assert sum(difficulty_counts.values()) == len(data['tasks'])
difficulty_counts
```

```python
count_to_tags = {}
for tag, count in tag_counts.items():
    count_to_tags.setdefault(count, []).append(tag)

rows = [(', '.join(sorted(tags)), count) for count, tags in sorted(count_to_tags.items())]
df = pd.DataFrame(rows, columns=['Tags', 'Number of Tasks'])
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
df
```

```python
plt.figure(figsize=(25, 8))
plt.bar(list(tag_counts.keys()), list(tag_counts.values()), color='skyblue')
plt.xlabel('Tags')
plt.ylabel('Number of tasks')
plt.title('ARCTIC-0 TIC tag frequencies')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(25, 8))
sns.barplot(x=list(tag_counts.keys()), y=list(tag_counts.values()), color='steelblue')
plt.xlabel('Tags')
plt.ylabel('Number of tasks')
plt.title('ARCTIC-0 TIC tag frequencies (seaborn)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
