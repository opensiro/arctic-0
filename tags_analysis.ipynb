```python
import json
from collections import Counter
import matplotlib.pyplot as plt
```

```python
# Load JSON from file
with open('dataset/arctic-0-85-0.7.0.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# Collect all tags from tasks
all_tags = []
for task in data.get('tasks', []):
    all_tags.extend(task.get('tags', []))

# Unique tags
unique_tags = set(all_tags)
print("Unique tags:", unique_tags)

# Frequency of each tag
tag_counts = Counter(all_tags)
print("Tag frequencies:")
for tag, count in tag_counts.items():
    print(f"{tag}: {count}")
```

```python
import pandas as pd
frequency_counts = Counter(tag_counts.values())

# --- Group tags by their count ---
count_to_tags = {}
for tag, count in tag_counts.items():
    count_to_tags.setdefault(count, []).append(tag)

# --- Convert to DataFrame for nice table ---
rows = []
for count, tags in sorted(count_to_tags.items()):
    rows.append([", ".join(tags), count])

df = pd.DataFrame(rows, columns=['Tags', 'Number of Tasks'])
pd.set_option('display.max_rows', None)       # показывать все строки
pd.set_option('display.max_colwidth', None)   # показывать всю ширину колонки

df
```

```python
plt.figure(figsize=(10, 6))
plt.bar(tag_counts.keys(), tag_counts.values(), color='skyblue')
plt.xlabel('Tags')
plt.ylabel('Number of tasks')
plt.title('Histogram of Tag Frequencies in Tasks')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
